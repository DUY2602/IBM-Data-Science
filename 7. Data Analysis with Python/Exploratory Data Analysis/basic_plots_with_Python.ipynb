# 🧠 Plotting Functions in Python

Data visualization helps us understand patterns and relationships in data.  
We’ll use **Matplotlib** and **Seaborn** to explore different types of plots in Python.

```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Enable inline plotting
%matplotlib inline
sns.set_theme(style="whitegrid")
```

## 📈 Line Plot

Line plots show how a variable changes over time or across continuous intervals.

```python
x = [1, 2, 3, 4, 5]
y = [3, 5, 7, 6, 9]

plt.figure(figsize=(6,4))
plt.plot(x, y, color='dodgerblue', marker='o', linestyle='--')
plt.title("Simple Line Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()
```

## 🎨 Customizing Plots

You can customize plots by adding:
- Titles and labels  
- Colors and line styles  
- Legends, grid lines, and annotations

```python
x = [1, 2, 3, 4, 5]
y1 = [3, 5, 7, 6, 9]
y2 = [2, 3, 4, 6, 8]

plt.figure(figsize=(7,4))
plt.plot(x, y1, label="Dataset A", color='tomato', marker='o')
plt.plot(x, y2, label="Dataset B", color='seagreen', marker='x')
plt.title("Customized Line Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.legend()
plt.grid(True)
plt.show()
```

## 📊 Bar and Pie Charts

Bar charts compare quantities among categories.  
Pie charts display proportions of a whole.

```python
categories = ['Apples', 'Bananas', 'Cherries', 'Dates']
values = [40, 25, 15, 20]

# Bar chart
plt.figure(figsize=(6,4))
plt.bar(categories, values, color=['#FF9999','#66B3FF','#99FF99','#FFCC99'])
plt.title("Fruit Sales Bar Chart")
plt.xlabel("Fruit Type")
plt.ylabel("Sales")
plt.show()

# Pie chart
plt.figure(figsize=(5,5))
plt.pie(values, labels=categories, autopct='%1.1f%%', startangle=90, colors=['#FF9999','#66B3FF','#99FF99','#FFCC99'])
plt.title("Fruit Sales Distribution")
plt.show()
```

## 🌈 Seaborn Visualizations

Seaborn provides high-level interfaces for statistical plots.  
We’ll use the built-in `tips` dataset for examples.

```python
tips = sns.load_dataset("tips")
tips.head()
```

```python
# Scatter plot with regression line
sns.lmplot(data=tips, x="total_bill", y="tip", hue="sex", height=5)
plt.title("Total Bill vs Tip by Gender")
plt.show()
```

```python
# Boxplot of total bill by day
plt.figure(figsize=(6,4))
sns.boxplot(data=tips, x="day", y="total_bill", palette="Set2")
plt.title("Boxplot of Total Bill by Day")
plt.show()
```

## 🧩 Subplots

Display multiple plots in one figure for comparison.

```python
x = [1, 2, 3, 4, 5]
y1 = [2, 4, 6, 8, 10]
y2 = [1, 3, 5, 7, 9]

plt.figure(figsize=(10,4))

plt.subplot(1, 2, 1)
plt.plot(x, y1, 'bo--')
plt.title("Line Plot")

plt.subplot(1, 2, 2)
plt.bar(x, y2, color='orange')
plt.title("Bar Plot")

plt.tight_layout()
plt.show()
```

## 🔥 Advanced: Pairplot and Heatmap

Pairplot shows pairwise relationships between variables.  
Heatmap visualizes correlations between numerical columns.

```python
# Pairplot
sns.pairplot(tips, hue="sex", palette="coolwarm")
plt.suptitle("Pairplot of Tips Dataset", y=1.02)
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(6,4))
sns.heatmap(tips.corr(numeric_only=True), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap (Tips Dataset)")
plt.show()
```
