```python
# Import the neccessary libraries and modules

import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
```

## Download & Import the Canadian immigratio dataset

```python
df = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/Canada.csv')

print("Data imported successfully")
```

**View the first 5 rows**

```python
df.head()
```

**Set the column Country as the index**

```python
df.set_index('Country', inplace=True)

# Verify the index
df.head()
```

**Declare a variable that will allow us to easily call upon the full range of years**

```python
# This variable will be very useful while plotting

years = list(map(str, range(1980, 2014)))
years
```

## Visualizing data using Matplotlib

```python
# Verify Matplotlib version
print('Matplotlib version: ', mpl.__version__)

# Get the available styling type
print(plt.style.available)

# Use ggplot for the project
mpl.style.use(['ggplot'])
```

## Line Plot

A line chart or line plot is a type of plot which displays information as a series of data points called 'markers' connected by straight line segments. It is a basic type of chart common in many fields. Use line plot when you have a continuous data set. These are best suited for trend-based visualizations of data over a period of time.

```python
vietnam = df.loc['Viet Nam', years]
print('The first 5 rows:\n ', vietnam.head())

# Give the plot a title
plt.title('Vietnam ---> Canada Immigration')

# Label for x-axis
plt.xlabel('Year')

# Label for y-axis
plt.ylabel('Number of Immigrants')

# Create the plot 
# Set the kind to 'line'
vietnam.index = vietnam.index.map(int)
vietnam.plot(kind='line')

x = vietnam.idxmax()
y = vietnam.max()

plt.text(x, y, f'Peak point with {y} immigrants in {x}')

plt.show()
```

## Practice

**Step 1:** Get the data set for China and India, and display the dataframe.

```python
df_CI = df.loc[['India', 'China'], years]
df_CI
```

**Step 2:** Plot graph. We will explicitly specify line plot by passing in kind parameter to plot().

```python
df_CI.plot(kind='line')
```

That doesn't look right...

Recall that pandas plots the indices on the x-axis and the columns as individual lines on the y-axis. Since df_CI is a dataframe with the country as the index and years as the columns, we must first transpose the dataframe using **transpose()** method to swap the row and columns.

```python
df_CI = df_CI.transpose()
df_CI
```

Plot the new transpose frame

```python
# let's change the index values of df_CI to type integer for plotting
df_CI.index = df_CI.index.map(int)
df_CI.plot(kind='line')

plt.title('Immigrants from China & India')
plt.xlabel('Years')
plt.ylabel('Number of Immigrants')

plt.show()
```

Compare the trend of **top 5 countries** that contributed the most to immigration to Canada.

```python
# Step 1: Get the dataset. 
# Recall that we created a Total column that calculates cumulative immigration by country. 
# We will sort on this column to get our top 5 countries using pandas sort_values() method.

df.sort_values(by='Total', ascending=False, axis=0, inplace=True)

# Get the top 5
df_top_5 = df.head()
# Transpose the dataframe
df_top_5 = df_top_5[years].transpose()

# View the dataframe
df_top_5

# Step 2: Plot the dataframe. 
# To make the plot more readeable, we will change the size using the `figsize` parameter.

# let's change the index values of df_top5 to type integer for plotting
df_top_5.index = df_top_5.index.map(int)

# pass a tuple (x, y) size
df_top_5.plot(kind='line', figsize=(14, 8))

plt.title('Immigration trend of the top 5 Countries')
plt.xlabel('Years')
plt.ylabel('Number of Immigrants')
```
